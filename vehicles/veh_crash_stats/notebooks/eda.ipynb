## Creating US Counties Choropleth Map using Holoviews

Let's create a choropleth map for 2020 population estimates

```python
con_public = ibis.postgres.connect(
    host='localhost',
    user='postgres',
    password='your_password',
    port=5432,
    database='analysis',
    schema='public',
)
```

```python
con_public.list_tables()
```

```python
us_pop2020 = con_public.table('us_counties_pop_est_2010_2020_raw')
```

```python
us_pop2020
```

#### We need the state and county FIPS codes in tuples because the bokeh library has counties geo shapes mapped to tuples

```python
from bokeh.sampledata.us_counties import data as counties
```

So a tuple of `(1,1)` represents Alabama's Autauga County.  The lat/lon for the polygons are then mapped to it.  See below.

```python
counties[(1,1)]
```

#### Converting the FIPS codes to tuples

First we need to get them from our ibis table expression, convert to list

```python
state_fips = us_pop2020.state.cast('int32').execute().values.tolist()
county_fips = us_pop2020.county.cast('int32').execute().values.tolist()
pop2020 = us_pop2020.popestimate2020.execute().values.tolist()
```

```python
state_fips[:10]
```

```python
county_fips[:10]
```

List of 2020 population estimates:

```python
pop2020[:10]
```

#### Now create tuples from the 2 lists using combination of `list()` and `zip()`

```python
state_county_fips = list(zip(state_fips, county_fips))
```

```python
state_county_fips[:10]
```

#### Now create a Python dictionary that maps the (state_fips, county_fips) tuple to their corresponding 2020 population estimates

```python
state_county_pop2020 = {key: value for key, value in zip(state_county_fips, pop2020)}
```

```python
state_county_pop2020[(1,1)]
```

#### We will get an error if the `counties` data set has a county that `state_county_pop2020` doesn't have and vice versa.  So we need to make sure we have counties that are present in both data sets.

```python
common_keys = list(counties.keys() & state_county_pop2020.keys())
```

```python
updated_counties = {key: counties[key] for key in counties.keys() if key in common_keys}
```

#### Now we finally have the geographical data prepred to make Holoviews choropleth maps

#### For a specific state

```python
chart_data = [dict(county, Population=state_county_pop2020[cid])
            for cid, county in updated_counties.items()
            if county["state"] == "al"]
```

```python
choropleth = hv.Polygons(chart_data, ['lons', 'lats'], [('detailed_name', 'County'), 'Population'])
```

```python
choropleth.opts(
    opts.Polygons(logz=True, tools=['hover'], xaxis=None, yaxis=None,
                   show_grid=False, show_frame=False, width=500, height=500,
                   color_index='Population', colorbar=True, toolbar='above', line_color='white'))
```

#### For all states

```python
chart_data = [dict(county, Population=state_county_pop2020[cid])
            for cid, county in updated_counties.items()]
```

```python
choropleth = hv.Polygons(chart_data, ['lons', 'lats'], [('detailed_name', 'County'), 'Population'])
```

```python
choropleth.opts(
    opts.Polygons(logz=True, tools=['hover'], xaxis=None, yaxis=None,
                   show_grid=False, show_frame=False, width=1700, height=400,
                   color_index='Population', colorbar=True, toolbar='above', line_color='white'))
```
